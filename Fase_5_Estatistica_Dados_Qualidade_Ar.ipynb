```python
# ==============================================================================
# ETAPA INICIAL DE PREPARAÇÃO E QUALIDADE DE DADOS
# ==============================================================================

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Leitura do arquivo original (especificando a aba de dados)
df = pd.read_excel('cidade_alfa_qualidade_ar_original.xlsx', sheet_name='Dados')

# Visualização inicial das primeiras linhas
df.head()
```

```python
# Quantidade de linhas e colunas disponíveis
df.shape

# Exibir os nomes das colunas disponíveis no dataframe
df.columns

# Exibir informações sobre os dados que se encontram no dataframe
df.info()

# Exibir informações de valores NULOS para cada coluna existente no dataframe
df.isnull().sum()

# Procurar por registros duplicados na base de dados
df.duplicated().sum()

# Exibir diversos dados estatísticos para cada coluna da base de dados
df.describe()
```

```python
# Exibir a quantidade de registros duplicados antes da remoção
df.duplicated().sum()

# Eliminar os registros duplicados existentes no dataframe
df = df.drop_duplicates()

# Exibir a quantidade de registros duplicados após a remoção
df.duplicated().sum()

# Exibindo a quantidade de conteúdos nulos que temos na coluna bairro
df['bairro'].isnull().sum()

# Substituindo o conteúdo nulo pelo texto "Não Informado"
df['bairro'] = df['bairro'].fillna('Não Informado')

# Exibindo a quantidade de conteúdos nulos que temos na coluna bairro após o tratamento
df['bairro'].isnull().sum()

# Exibindo a quantidade de ocorrências agrupadas por bairro
df['bairro'].value_counts(dropna=False)

# Filtrando valores inconsistentes de temperatura (mantendo o intervalo realista)
df = df[(df['temperatura_c'] >= -10) & (df['temperatura_c'] <= 50)]
```

```python
# ETAPA DE ANÁLISE ESTATÍSTICA
# Realizando a coleta de estatísticas descritivas para as variáveis de interesse
df[['pm25_ug_m3', 'temperatura_c']].describe()

# Gerando a correlação entre a temperatura ambiente e a concentração de PM2.5 na Cidade Alfa
correlacao = df['pm25_ug_m3'].corr(df['temperatura_c'])

# Exibindo o resultado da correlação entre as 2 variáveis
print("Coeficiente de Correlação:", correlacao)
```

```python
# Criação do Dicionário de Dados para a segunda aba
df_dicionario = pd.DataFrame({
    "Coluna": [
        "id_coleta", "id_ponto_monitoramento", "data_coleta", "regiao", 
        "bairro", "tipo_area", "temperatura_c", "umidade_%", 
        "velocidade_vento_kmh", "chuva_mm", "pm25_ug_m3", "pm10_ug_m3", 
        "co_ppm", "no2_ppb", "o3_ppb", "indice_qualidade_ar", "qualidade_percebida"
    ],
    "Descrição": [
        "identificador único da medição.",
        "local onde o sensor está instalado.",
        "data/hora da medição.",
        "região da cidade.",
        "bairro do ponto de monitoramento.",
        "residencial, industrial, comercial, escolar etc.",
        "temperatura em °C.",
        "umidade relativa do ar (%).",
        "velocidade do vento (km/h).",
        "precipitação registrada (mm).",
        "concentração de material particulado fino PM2,5 (µg/m³).",
        "concentração de material particulado PM10 (µg/m³).",
        "concentração de monóxido de carbono (ppm).",
        "concentração de dióxido de nitrogênio (ppb).",
        "concentração de ozônio (ppb).",
        "índice calculado a partir dos poluentes.",
        "percepção da população: Boa, Moderada, Ruim, Péssima etc."
    ]
})

# Salvando a planilha Excel contendo as abas 'Dados' e 'Dicionario'
with pd.ExcelWriter("cidade_alfa_qualidade_ar_ajustado.xlsx", engine='openpyxl') as writer:
    df.to_excel(writer, sheet_name='Dados', index=False)
    df_dicionario.to_excel(writer, sheet_name='Dicionario', index=False)

print("Arquivo gerado com sucesso!")
```
